```python
# рассчитаем границы для выбросов
Q1 = np.percentile(payments['sum_payment'], 25)
Q3 = np.percentile(payments['sum_payment'], 75)
IQR = Q3 - Q1
lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR
print(lower_bound)
print(upper_bound)
```

```python
payments_clean = payments.loc[payments['sum_payment']< upper_bound].copy()
payments_clean.head()
```

### 2. Агрегация данных

```python
# создадим колонку год+месяц
payments_clean['mon'] = payments_clean['date'].dt.strftime('%Y-%m')
payments_clean.head()
```

```python
pivot_pay  = pd.pivot_table(payments_clean,\
                            values='sum_payment', \
                            index=['id_user', 'mon'],\
                            columns='training_type',\
                            aggfunc=['count', 'sum'],\
                            fill_value=0).reset_index()
pivot_pay.head()
```

```python
pivot_pay.columns = ['id_user', 'mon', 'cnt_group', 'cnt_indiv', 'sum_group', 'sum_indiv']
pivot_pay['cnt_total'] = pivot_pay['cnt_group']+pivot_pay['cnt_indiv']
pivot_pay['sum_total'] = pivot_pay['sum_group']+pivot_pay['sum_indiv']
pivot_pay.head()
```

### 3. Анализ выручки и количества тренировок

```python
p_gr = pivot_pay.groupby('mon')[['cnt_group', 'cnt_indiv', 'cnt_total', 'sum_group', 'sum_indiv', 'sum_total']].\
sum().reset_index()
p_gr.head()
```

```python
# динамика тренировок
plt.figure(figsize=(8, 5))
plt.plot(p_gr['mon'], p_gr['cnt_group'], marker='o', label='Групповые')
plt.plot(p_gr['mon'], p_gr['cnt_indiv'], marker='o', label='Индивидуальные')
plt.plot(p_gr['mon'], p_gr['cnt_total'], marker='o', label='Всего')
plt.xlabel('Месяц')
plt.ylabel('Количество тренировок')
plt.title('Тренировки в динамике')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)  # начинаем ось Y с нуля
plt.xticks(rotation=45)  # поворот подписей на оси X
plt.show()
```

```python
# динамика выручки
plt.figure(figsize=(8, 5))
plt.plot(p_gr['mon'], p_gr['sum_group'], marker='o', label='Групповые')
plt.plot(p_gr['mon'], p_gr['sum_indiv'], marker='o', label='Индивидуальные')
plt.plot(p_gr['mon'], p_gr['sum_total'], marker='o', label='Всего')
plt.xlabel('Месяц')
plt.ylabel('Выручка')
plt.title('Выручка в динамике')
plt.legend()
plt.grid(True)
plt.ylim(bottom=0)  # начинаем ось Y с нуля
plt.xticks(rotation=45)  # поворот подписей на оси X
plt.show()
```

### 4. Считаем LT и LTR

```python
# добавим колонки с первым и послденим месяцем занятий
pivot_pay['min_mon'] = pivot_pay.groupby('id_user')['mon'].transform('min')
pivot_pay['max_mon'] = pivot_pay.groupby('id_user')['mon'].transform('max')
pivot_pay.head()
```

```python
# рассчитаем средний и медианный LTR
pay_ltr = pivot_pay.groupby('id_user').agg(ltr = ('sum_total', 'sum')).reset_index()
print(round(pay_ltr['ltr'].mean(), 2))
print(round(pay_ltr['ltr'].median(), 2))
```

```python
# распределение LTR
sns.histplot(pay_ltr['ltr'], kde = True)
```

```python
# отбросим часть данных, где max_mon = декабрь 2023
clean_pay = pivot_pay.loc[pivot_pay['max_mon']!='2023-12']
print(len(pivot_pay))
print(len(clean_pay))
```

```python
# рассчитаем ltr на оставшихся данных
clean_gr = clean_pay.groupby('id_user').agg(ltr = ('sum_total', 'sum')).reset_index()
```

```python
print(round(clean_gr['ltr'].mean(), 2))
print(round(clean_gr['ltr'].median(), 2))
```

```python
# теперь отфильтруем данные по min_mon и оставим самые ранние когорты
clean_pay2 = pivot_pay.loc[pivot_pay['min_mon'].isin(['2023-01', '2023-02', '2023-03', '2023-04'])]
print(len(pivot_pay))
print(len(clean_pay2))
```

```python
#снова посчитаем LTR
clean_gr2 = clean_pay2.groupby('id_user').agg(ltr = ('sum_total', 'sum')).reset_index()
```

```python
print(round(clean_gr2['ltr'].mean(), 2))
print(round(clean_gr2['ltr'].median(), 2))
```

```python
#рассчитаем LT
p_group = clean_pay2.groupby('id_user').agg(nmon = ('id_user', 'count')).reset_index()
print(p_group['nmon'].mean())
print(p_group['nmon'].median())
```

```python
print(f"среднее количество месяцев посещений фитнес-клуба: {round(p_group['nmon'].mean(), 2)} месяцев")
```

**Часть 2. Демография и поведенческая сегментация:** Анализ региональной структуры клиентов по полу и городам, а также кластеризация пользователей по уровню их активности.

```python
users_info = pd.read_csv('users_info.csv')
users_info.head()
```

### EDA

```python
#Кол-во строк и столбцов
users_info.shape
```

```python
#Типы данных
users_info.dtypes
```

```python
#Кол-во пропущенных значений 
users_info.isnull().sum()
```

```python
#Удаление пропусков, Процент удаленнения 
len_us=len(users_info)
users_info = users_info.dropna(subset=['age', 'gender'])
per_u = round((len_us- len(users_info)) / len_us* 100, 2)
per_u
```

```python
#Выбросы по возврасту 
Q1 = np.percentile(users_info['age'], 25)
Q3 = np.percentile(users_info['age'], 75)
IQR = Q3 - Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
print('Нижняя граница', lower_bound, 'Верхняя граница', upper_bound)
```

```python
#Удаление выбросов
users_clean = users_info[(users_info['age'] >= lower_bound)]
users_clean.head()
```

```python
#Визулизация 
sns.histplot(users_clean['age'], alpha = 0.5, color='green')
plt.title('Распределение возрастов пользователей')
plt.xlabel('Возраст')
plt.ylabel('Количество пользователей')
plt.show()
```

```python
#Рассчет количества пользователей по каждому полу и городу (с помощью группировки и агрегации)

cnt_us=users_clean.groupby(['gender', 'city']).agg(cnt_us=('id_user', 'count'))
cnt_us.head()
```

```python
#Рассчет количества пользователей по каждому полу и городу (Группировка pivot_tabel)
pivot_us = users_clean.pivot_table(
    index='city',     # строки 
    columns='gender',      # столбцы 
    values='id_user',         # агрегируемая величина
    aggfunc= 'count',     # функция агрегации
    fill_value=0                  # чем заполнить пропуски
).reset_index()

pivot_us.columns=['city', 'cnt_f', 'cnt_m']
pivot_us.head()
```

```python
#Доля женщин среди всех пользователей
pivot_us['all_us'] = pivot_us['cnt_f'] + pivot_us['cnt_m']
pivot_us['f_rate'] = round(pivot_us['cnt_f'] / pivot_us['all_us'], 2)
pivot_us
```

```python
#Визуализация
import matplotlib.pyplot as plt

pivot_us.plot(
    kind='bar', 
    x='city', 
    y='f_rate', 
    color='skyblue', alpha = .9,
    legend=False)
plt.title('Доля женщин по городам')
plt.xlabel('Город')
plt.ylabel('Доля')
plt.xticks(rotation=45) 
plt.show()
```

```python
# Разделение пользователей на бины 

pay_mon = pd.read_csv('payments_monthly.csv')
cnt_pay=pay_mon.groupby('id_user').agg(avg_tr=('cnt_total', 'mean')).reset_index()
cnt_pay
```

```python
#  Разделение пользователей на группы по активности

bins = [0, 6, 10, float('inf')]
labels = ['мало', 'средне', 'много']
cnt_pay['group'] =pd.cut(cnt_pay['avg_tr'], bins=bins, labels=labels, right=True)
cnt_gr = cnt_pay['group'].value_counts()
cnt_gr
```
